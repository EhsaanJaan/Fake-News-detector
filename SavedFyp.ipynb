```python
import joblib
import gradio as gr

##loadng the entire pipeine of the fake news detector through the pickle and then launching the ui
loaded_pipeline = joblib.load('fyp_welfake_ensemble_model.pkl')

def predict_news(text):
    probs = loaded_pipeline.predict_proba([text])[0]
    
    return {"Real News": float(probs[0]), "Fake News": float(probs[1])}

interface = gr.Interface(
    fn=predict_news,
    inputs=gr.Textbox(lines=10, placeholder="Paste a news article here to test your ensemble..."),
    outputs=gr.Label(num_top_classes=2),
    title="200143671 Fake News Detector",
    description="Made with a Soft-Voting Ensemble (Logistic Regression, Calibrated SVC, LightGBM). Percentages are not the chance of article being real or fake, but a display the average confidence scores of the models"
)

interface.launch()
```
